```python
from statsmodels.tsa.stattools import adfuller

result = adfuller(ts)
print("ADF Statistics: ", result[0])
print("p- value : ", result[1])
```

```python
ts_diff =ts.diff().dropna()
```

```python
# from statsmodels.tsa.stattools import adfuller

result = adfuller(ts_diff)
print("ADF Statistics: ", result[0])
print("p- value : ", result[1])

# p-value = 0 means it is stationary series
```

```python
# . What the Augmented Dickey-Fuller (ADF) Test Does
# - The ADF test checks whether a time series has a unit root.
# - Null hypothesis (H₀): The series has a unit root → it is non-stationary.
# - Alternative hypothesis (H₁): The series is stationary (no unit root).
# - Stationarity means the mean, variance, and autocorrelation structure of the series do not change over time.
# Stationary means p-value close to 0 and non unit root


```

```python
from statsmodels.tsa.arima.model import ARIMA
```

```python
model = ARIMA(ts_diff, order = (2,1,1))
model_fit = model.fit()
print(model_fit.summary())
```
